# **4. Calculate the sample mean of the variables: count_rental_bikes, casual_users, registered_users.**

# **8. Create a scatter plot of the relationship between norm_temperature (X) and count_rental_bikes (Y).**

````python
plt.figure(figsize=(7,4)) # Create space for the plot.
sns.scatterplot(data = bike_data, x = 'norm_temperature', y = 'count_rental_bikes') # Show the plot.
plt.title("Scatter plot of Norm Temperature vs Rental Bikes") # Set plot title.
plt.xlabel("Norm Temperature") # Set label for X axis.
plt.ylabel("Rental Bikes") # Set label for Y axis.
plt.show() # Show plot.
````

**Interpret the plot.**

Answer: Al observar la gráfica de dispersión entre la variable norm_temperature y count_rental_bikes, se aprecia una tendencia positiva en la que el número de bicicletas rentadas aumenta conforme la temperatura se eleva, hasta llegar a un punto en el que la relación comienza a estabilizarse o incluso decrecer. Este comportamiento refleja que existe un rango óptimo de temperatura en el cual la demanda de bicicletas alcanza su máximo, mientras que temperaturas extremas reducen el uso. La dispersión de los puntos también evidencia que, aunque la temperatura es un factor relevante, no es el único que influye en las rentas, ya que otros elementos como condiciones climáticas adicionales o factores externos pueden explicar las variaciones observadas.



# **9. Standardized all numerical variables so they are on the same scale.**

````python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
````

````python
#Standarization of predictors
bike_data_Num=(bike_data
        .filter(["norm_temperature","norm_humidity","norm_windspeed",
                 "casual_users","registered_users","count_rental_bikes"]))

scaler = StandardScaler()
bike_data_Num_scaled= scaler.fit_transform(bike_data_Num)
bike_data_Num_scaled
````

````python
#Verifying the scaling
bike_data_Num_scaled = pd.DataFrame(bike_data_Num_scaled, columns = bike_data_Num.columns)
bike_data_Num_scaled
````

# **10. Construct a color map on correlation for the standardized numerical variables. Identify the pairs of predictors that are highly correlated (if any). Use a threshold of 0.75.**

````python
correlation_matrix = bike_data_Num_scaled.corr()
print(correlation_matrix)
````

````python
#plot
plt.figure(figsize=(7,3))
sns.heatmap(correlation_matrix, cmap='coolwarm', annot = True)
````

En este caso las variables de registered_users y count_rental_bikes tienen una correlación alta de 0.95, por lo que se procederá a realizar un scatter plot entre ambas para ver su comportamiento:


````python
plt.figure(figsize=(7,4)) # Create space for the plot.
sns.scatterplot(data = bike_data_Num_scaled, x = 'registered_users', y = 'count_rental_bikes', color="red") # Show the plot.
plt.title("Scatter plot of Registered Users vs Rental bikes") # Set plot title.
plt.xlabel("Registered Users") # Set label for X axis.
plt.ylabel("Rental Bikes") # Set label for Y axis.
plt.show() # Show plot.
````

**Interpret the plot.**

Answer: La gráfica de dispersión entre registered_users y count_rental_bikes muestra una clara relación positiva: conforme aumenta el número de usuarios registrados, también se incrementa el número de bicicletas rentadas. Este patrón indica que los usuarios registrados son un factor determinante en la demanda del servicio, ya que su participación está directamente asociada con un mayor volumen de rentas.
